# AI Image Restoration & Colorization
This notebook demonstrates how to restore and colorize old black and white photos using Deep Learning and OpenCV.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from PIL import Image
from utils import load_colorizer, colorize_image, restore_image

# Ensure models are downloaded
net = load_colorizer()
```

## Load an Image
Upload or specify a path to an old B&W image.

```python
# For demo, we'll use a public image URL if no local file is provided
import requests
from io import BytesIO

url = "https://raw.githubusercontent.com/richzhang/colorization/master/demo/imgs/ansel_adams3.jpg"
response = requests.get(url)
img = Image.open(BytesIO(response.content)).convert("RGB")
image_np = np.array(img)

plt.figure(figsize=(10, 5))
plt.imshow(image_np)
plt.title("Original B&W Image")
plt.axis('off')
plt.show()
```

## Apply Restoration
Denoising and sharpening to improve quality.

```python
restored = restore_image(image_np)

plt.figure(figsize=(10, 5))
plt.imshow(restored)
plt.title("Restored (Denoised & Sharpened)")
plt.axis('off')
plt.show()
```

## Apply AI Colorization
Using the Caffe model to predict colors.

```python
colorized = colorize_image(restored, net)

plt.figure(figsize=(10, 5))
plt.imshow(colorized)
plt.title("Final Colorized Image")
plt.axis('off')
plt.show()
```

## Comparison Table

```python
fig, axes = plt.subplots(1, 3, figsize=(20, 10))
axes[0].imshow(image_np)
axes[0].set_title("Original B&W")
axes[0].axis('off')

axes[1].imshow(restored)
axes[1].set_title("Restored")
axes[1].axis('off')

axes[2].imshow(colorized)
axes[2].set_title("Colorized")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
